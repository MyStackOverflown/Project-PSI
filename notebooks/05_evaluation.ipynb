## Ewaluacja i generacja tekstu

### Import zależności

```python
from src.data.dataset import load_vocab_artifacts
from src.models.generate import build_inference_model, generate_text
from src.evaluation.metrics import (
    history_to_metrics,
    evaluate_generation,
    save_metrics
)
from src.utils.paths import PROCESSED_DATA_DIR, FINAL_MODELS_DIR, METRICS_DIR, PROCESSED_DATA_DIR
```

### Wczytanie słownika znaków

```python
vocab, char2idx, idx2char = load_vocab_artifacts(PROCESSED_DATA_DIR)
```

### Budowa modelu inferencyjnego

```python
model = build_inference_model(
    vocab_size=len(vocab),
    embedding_dim=512,
    rnn_units=2048,
    weights_path=FINAL_MODELS_DIR / "asoiaf_10_512_2048.keras"
)
```

### Generacja tekstu

```python
text = generate_text(
    model,
    start_string="Robert Baratheon",
    char2idx=char2idx,
    idx2char=idx2char,
    temperature=0.1
)
```

### Ewaluacja generacji

```python
metrics = evaluate_generation(
    source_text=(PROCESSED_DATA_DIR / "asoiaf.txt").read_text(encoding="utf-8"),
    generated_text=text
)
```

### Zapis metryk

```python
save_metrics(metrics, METRICS_DIR / "generation_metrics_10_512_2048.json")
```
